> For demonstration and testing purposes only — the regulatory corpus is synthetic and the sample submission is fictional. See `DISCLAIMER.md`.

**Status**: planned. For now, run the CLI from the repository root: `python -m src.main --submission data/sample_submission/sample_package.json`.
